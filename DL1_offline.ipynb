```python
# Assignment 1 - quick demonstration of the Keras / TensorFlow packages
import tensorflow as tf
import keras
print("TensorFlow version:", tf.__version__)
print("Keras version     :", keras.__version__)
print("Devices           :", [d.device_type for d in tf.config.list_physical_devices()])

# TensorFlow: tensors and automatic differentiation
a = tf.constant([[1., 2.], [3., 4.]])
b = tf.constant([[5., 6.], [7., 8.]])
print("\nTensor addition:\n", (a + b).numpy())
print("Matrix multiplication:\n", tf.matmul(a, b).numpy())
x = tf.Variable(3.0)
with tf.GradientTape() as tape:
    y = x ** 2 + 2 * x
print("dy/dx of x^2 + 2x at x=3 :", tape.gradient(y, x).numpy())

# Keras: define -> compile -> fit -> evaluate -> predict
import numpy as np
np.random.seed(0)
X = np.random.rand(500, 4).astype("float32")
Y = (X.sum(axis=1) > 2).astype("float32")
model = keras.Sequential([keras.layers.Input((4,)),
                          keras.layers.Dense(8, activation="relu"),
                          keras.layers.Dense(1, activation="sigmoid")])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
model.fit(X, Y, epochs=15, batch_size=32, verbose=0)
print("Evaluate [loss, accuracy]:", [round(float(v), 4) for v in model.evaluate(X, Y,
    verbose=0)])
print("Prediction for [0.9,0.8,0.7,0.6]:", model.predict(np.array([[0.9, 0.8, 0.7, 0.6]]),
    verbose=0).round(3))
```
